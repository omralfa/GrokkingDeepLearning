## 1. Ringkasan Teori

### A. Supervised dan Unsupervised Learning

**Supervised learning** mempelajari pola dari data masukan yang disertai target. Model membuat prediksi, kemudian hasilnya dibandingkan dengan target yang diketahui.

**Unsupervised learning** mempelajari struktur dari data tanpa target yang disediakan. Salah satu contohnya adalah mengelompokkan data berdasarkan kemiripan.

### B. Model Parametrik dan Nonparametrik

Model **parametrik** memiliki jumlah parameter yang tetap. Pembelajaran dilakukan dengan menyesuaikan nilai parameter tersebut.

Model **nonparametrik** memungkinkan jumlah parameter atau representasi yang dipakai bergantung pada data. Buku menjelaskan perbedaan ini secara konseptual melalui pendekatan penyesuaian parameter dan penghitungan pola.

Kedua pengelompokan tersebut berbeda: supervised/unsupervised menjelaskan jenis pola yang dipelajari, sedangkan parametrik/nonparametrik menjelaskan bagaimana pembelajaran direpresentasikan.

### C. Predict, Compare, Learn

Buku memperkenalkan tiga tahap dasar pembelajaran:

1. **Predict:** model menghasilkan prediksi dari masukan.
2. **Compare:** prediksi dibandingkan dengan target untuk mengetahui kesalahannya.
3. **Learn:** model menggunakan informasi kesalahan untuk memperbaiki prediksi berikutnya.

Kerangka ini menjadi dasar untuk memahami implementasi jaringan saraf pada bab berikutnya.

## 2. Reproduksi Kode dan Ilustrasi

Bab ini terutama menjelaskan konsep dasar. Sel berikut adalah ilustrasi sederhana dari pola *predict, compare, learn*, bukan klaim bahwa kode ini disalin dari buku.

Kita menggunakan model linear sederhana dengan satu parameter. Target diketahui, prediksi dihitung, kesalahan diukur, lalu parameter diperbarui secara bertahap.


```python
# Ilustrasi konsep predict, compare, learn
# Contoh ini dibuat untuk memahami konsep, bukan salinan kode asli buku.

target = 8.0
input_value = 2.0
weight = 1.0
learning_rate = 0.1

print("Bobot awal:", weight)

for step in range(5):
    # Predict
    prediction = input_value * weight

    # Compare
    error = target - prediction

    # Learn: pembaruan sederhana berdasarkan error
    weight += learning_rate * error / input_value

    print(
        f"Langkah {step + 1}: "
        f"prediksi={prediction:.2f}, "
        f"error={error:.2f}, "
        f"bobot baru={weight:.2f}"
    )

print("Prediksi akhir:", input_value * weight)
```

## 3. Analisis Hasil

Pada contoh ilustratif, prediksi awal belum sama dengan target. Selisih antara target dan prediksi digunakan untuk memperbarui bobot.

Setiap pembaruan mengubah prediksi berikutnya agar lebih dekat ke target. Contoh ini menunjukkan gagasan dasar pembelajaran melalui kesalahan, tetapi bukan implementasi lengkap algoritma pelatihan jaringan saraf.

Supervised learning menggunakan target untuk membandingkan prediksi, sedangkan unsupervised learning mencari struktur tanpa target yang telah ditentukan. Model parametrik dan nonparametrik dibedakan berdasarkan cara representasi pembelajaran ditentukan.
