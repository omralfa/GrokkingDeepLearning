## 1. Tujuan Bab

Bab ini menjelaskan bagaimana jaringan saraf belajar menggunakan beberapa bobot sekaligus. Setiap input memiliki bobot masing-masing, tetapi semua bobot berkontribusi terhadap satu prediksi.

## 2. Konsep Teori

### A. Prediksi dengan Banyak Input

Untuk tiga input, prediksi linear dapat ditulis:

$$
\hat{y}=x_1w_1+x_2w_2+x_3w_3
$$

Setiap pasangan input dan bobot berkontribusi terhadap hasil prediksi.

### B. Error dan Delta

Squared error mengukur perbedaan prediksi dengan target:

$$
E=(\hat{y}-y)^2
$$

Dalam contoh pembelajaran pada bab ini, delta didefinisikan sebagai:

$$
\delta=\hat{y}-y
$$

Delta digunakan untuk menentukan arah pembaruan bobot.

### C. Weight Deltas

Untuk model linear dengan satu output, kontribusi setiap input terhadap pembaruan bobot dihitung dari delta dikalikan input terkait:

$$
\Delta w_i=\delta x_i
$$

### D. Pembaruan Bobot

Bobot diperbarui menggunakan learning rate atau alpha:

$$
w_i \leftarrow w_i-\alpha\Delta w_i
$$

Dengan demikian, setiap bobot memperoleh pembaruan sesuai input yang terhubung dengannya.

## 3. Implementasi Ilustratif

Kode berikut mendemonstrasikan konsep tersebut. Ini merupakan contoh mandiri, bukan klaim salinan kode asli buku.

```python
# Prediksi dengan tiga input dan tiga bobot

def neural_network(inputs, weights):
    return sum(x * w for x, w in zip(inputs, weights))

inputs = [8.5, 0.65, 1.2]
target = 1.0
weights = [0.1, 0.2, -0.1]

prediction = neural_network(inputs, weights)
error = (prediction - target) ** 2
delta = prediction - target

weight_deltas = [delta * x for x in inputs]

print("Prediksi awal:", prediction)
print("Target:", target)
print("Squared error:", error)
print("Delta:", delta)
print("Weight deltas:", weight_deltas)
```

```python
# Perbarui setiap bobot satu kali

alpha = 0.01

updated_weights = [
    w - alpha * wd
    for w, wd in zip(weights, weight_deltas)
]

print("Bobot sebelum:", weights)
print("Bobot sesudah:", updated_weights)
print("Prediksi sesudah:", neural_network(inputs, updated_weights))
```

## 4. Analisis Hasil

Prediksi dihitung dari jumlah perkalian setiap input dengan bobotnya. Delta mengukur selisih prediksi dan target, sedangkan weight deltas menunjukkan kontribusi masing-masing input terhadap pembaruan bobot dalam contoh ini.

Setelah bobot diperbarui, prediksi dapat berubah. Satu pembaruan belum menjamin error menjadi minimum; proses pembelajaran biasanya memerlukan iterasi dan pengaturan learning rate.

**Batasan:** implementasi ini menggunakan model linear satu output dan tidak mencakup seluruh variasi pembelajaran jaringan saraf.
