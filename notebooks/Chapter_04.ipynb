## 1. Tujuan Bab

Bab ini membahas bagaimana jaringan saraf belajar dengan mengurangi kesalahan prediksi. Proses pembelajaran dijelaskan melalui perbandingan prediksi dengan target dan penyesuaian bobot.

## 2. Konsep Teori

### A. Predict, Compare, and Learn

Pembelajaran jaringan saraf dapat dipahami melalui tiga tahap:

1. **Predict:** model menghasilkan prediksi dari input.
2. **Compare:** prediksi dibandingkan dengan nilai target untuk mengukur kesalahan.
3. **Learn:** bobot diperbarui agar kesalahan dapat berkurang.

### B. Mengukur Error

Salah satu ukuran kesalahan yang digunakan dalam pembelajaran adalah squared error:

\[
E = (\hat{y}-y)^2
\]

Keterangan:
- \(\hat{y}\): prediksi model;
- \(y\): target sebenarnya;
- \(E\): squared error.

Semakin kecil error, semakin dekat prediksi terhadap target pada contoh tersebut.

### C. Hot and Cold Learning

Pendekatan hot and cold learning menguji perubahan bobot ke arah berbeda dan membandingkan hasil error-nya. Pendekatan ini membantu memahami bagaimana perubahan bobot memengaruhi prediksi, tetapi pengujian manual dapat menjadi tidak efisien.

### D. Gradient Descent

Gradient descent memperbarui bobot berdasarkan arah yang mengurangi error. Secara umum:

\[
w_{baru}=w_{lama}-\eta\frac{\partial E}{\partial w}
\]

\(\eta\) merupakan learning rate yang mengatur besar langkah pembaruan.

## 3. Eksperimen Ilustratif

Kode berikut adalah ilustrasi sederhana gradient descent untuk satu input dan satu bobot, bukan salinan kode asli buku.

```python
# Ilustrasi gradient descent untuk model prediksi y_hat = x * w

x = 2.0
target = 8.0
weight = 1.0
learning_rate = 0.1

for step in range(10):
    prediction = x * weight
    error = (prediction - target) ** 2

    # Turunan squared error terhadap weight
    gradient = 2 * (prediction - target) * x

    weight -= learning_rate * gradient

    print(
        f"Langkah {step + 1:2d} | "
        f"Prediksi: {prediction:.4f} | "
        f"Error: {error:.4f} | "
        f"Bobot baru: {weight:.4f}"
    )

print("Prediksi akhir:", x * weight)
print("Target:", target)
```

## 4. Analisis Hasil

Perhatikan perubahan error dan bobot pada setiap langkah. Gradient descent menggunakan gradien untuk menentukan arah pembaruan bobot.

Pada contoh ini, model bersifat linear dengan satu input. Hasilnya tidak dapat langsung digeneralisasi ke seluruh jaringan saraf karena jaringan yang lebih kompleks memiliki banyak bobot dan fungsi yang berbeda.
