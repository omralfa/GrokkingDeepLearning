## Tujuan bab
Bab ini membahas fungsi aktivasi, khususnya **sigmoid**, **tanh**, dan **softmax**, serta peran nonlinieritas dalam jaringan saraf.

## Ringkasan teori dari buku
Fungsi aktivasi diterapkan pada keluaran neuron saat prediksi. ReLU yang digunakan pada bab sebelumnya mengubah nilai negatif menjadi nol. Fungsi aktivasi perlu menghasilkan keluaran yang terdefinisi untuk masukan yang relevan; pemilihan fungsi juga bergantung pada tujuan lapisan.

- **Sigmoid:** memetakan nilai riil ke rentang (0, 1), sering digunakan untuk memodelkan probabilitas biner.
- **Tanh:** memetakan nilai ke rentang (-1, 1), dengan titik tengah nol.
- **Softmax:** mengubah sekumpulan skor menjadi distribusi probabilitas yang jumlahnya satu, cocok untuk kelas-kelas yang saling eksklusif.
- **Nonlinieritas:** tanpa fungsi aktivasi nonlinier, susunan beberapa transformasi linear tetap ekuivalen dengan satu transformasi linear; kedalaman saja tidak cukup untuk memodelkan hubungan kompleks.

Contoh di bawah adalah implementasi NumPy untuk memeriksa sifat fungsi tersebut.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -40, 40)))

def tanh(x):
    return np.tanh(x)

def softmax(scores):
    # Mengurangi maksimum untuk stabilitas numerik.
    shifted = scores - np.max(scores, axis=-1, keepdims=True)
    exp_scores = np.exp(shifted)
    return exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

x = np.array([-3.0, -1.0, 0.0, 1.0, 3.0])
print("x      :", x)
print("sigmoid:", np.round(sigmoid(x), 4))
print("tanh   :", np.round(tanh(x), 4))

scores = np.array([[2.0, 1.0, 0.1],
                   [0.5, 2.5, 1.0]])
probabilities = softmax(scores)
print("Softmax:\n", np.round(probabilities, 4))
print("Jumlah probabilitas per baris:", probabilities.sum(axis=1))
```

## Memilih fungsi aktivasi
| Fungsi | Rentang keluaran | Penggunaan umum |
|---|---:|---|
| ReLU | [0, ∞) | Lapisan tersembunyi |

| Tanh | (-1, 1) | Lapisan tersembunyi pada arsitektur tertentu |
| Softmax | Tiap kelas 0–1; jumlah 1 | Klasifikasi multi-kelas |

Tabel ini adalah ringkasan penggunaan umum; pilihan aktual bergantung pada arsitektur, loss function, dan karakteristik data.

## Kesimpulan
Fungsi aktivasi memberi jaringan kemampuan memodelkan hubungan nonlinier dan menentukan cara menafsirkan keluaran neuron. Untuk klasifikasi multi-kelas, softmax menghasilkan distribusi probabilitas yang dapat dibandingkan dengan label target.
