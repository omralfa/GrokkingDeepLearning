## 1. Ringkasan Teori

### A. Mengapa mempelajari deep learning?

Deep learning merupakan bagian dari perkembangan machine learning dan kecerdasan buatan yang bertujuan mengotomatisasi kemampuan untuk mempelajari pola dan menyelesaikan tugas. Buku ini menyoroti keberhasilannya dalam bidang computer vision, pengenalan suara, dan penerjemahan mesin.

Deep learning juga berpotensi mengotomatisasi sebagian pekerjaan yang membutuhkan keterampilan manusia. Selain manfaat praktisnya, mempelajari cara mesin meniru aspek kecerdasan dapat membantu kita memahami proses belajar, penalaran, dan kreativitas.

### B. Mengapa membaca buku ini?

Buku ini dirancang agar pembaca dapat mempelajari deep learning tanpa harus memiliki latar belakang matematika tingkat lanjut. Konsep dijelaskan secara bertahap melalui teori dan praktik kode, sehingga pembaca tidak hanya memahami istilah, tetapi juga mempelajari cara kerja komponen jaringan saraf.

Pendekatan buku ini adalah membangun pemahaman melalui implementasi, kemudian menjelaskan konsep yang mendasarinya. Dengan demikian, pembaca dapat memahami bagaimana dan kapan teknik tertentu digunakan.

### C. Persiapan yang dibutuhkan

Menurut buku, perangkat utama yang digunakan adalah:

- **Python:** bahasa pemrograman untuk menjalankan contoh.
- **Jupyter Notebook:** lingkungan interaktif untuk menjalankan kode dan memeriksa hasil setiap langkah.
- **NumPy:** pustaka operasi numerik dan matriks yang digunakan untuk membangun contoh deep learning dari dasar.
- **Matematika dasar:** pemahaman aljabar dasar sudah menjadi titik awal yang cukup.

Buku juga menyarankan pembaca memilih masalah yang menarik bagi dirinya agar proses belajar lebih terarah dan relevan.
