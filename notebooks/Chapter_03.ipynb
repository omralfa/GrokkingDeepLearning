## 1. Tujuan Bab

Bab ini memperkenalkan tahap prediksi (*predict*) pada jaringan saraf. Prediksi adalah proses ketika model menerima data masukan dan menghasilkan keluaran berdasarkan bobot yang dimilikinya.

Pembahasan dimulai dari satu input dan satu output, kemudian berkembang menuju beberapa input dan output. Jumlah dan bentuk data yang dimasukkan perlu disesuaikan dengan informasi yang dibutuhkan untuk menyelesaikan masalah.

## 2. Konsep Teori

### A. Prediksi (*Prediction*)

Jaringan saraf menerima data, memprosesnya menggunakan bobot, kemudian menghasilkan prediksi. Pada tahap ini, model belum tentu menghasilkan prediksi yang benar.

### B. Bobot (*Weight*)

Bobot merupakan parameter yang mengatur pengaruh input terhadap output. Dalam jaringan paling sederhana dengan satu input dan satu bobot, prediksi dapat dinyatakan sebagai:

\[
\hat{y} = xw
\]

dengan:
- \(x\): nilai input;
- \(w\): bobot;
- \(\hat{y}\): hasil prediksi.

### C. Bentuk Input dan Output

Model harus menerima informasi yang cukup untuk menghasilkan prediksi. Misalnya, memprediksi keberadaan kucing dalam gambar memerlukan informasi visual yang memadai, bukan sekadar satu piksel.

Jumlah input dan output yang digunakan bergantung pada permasalahan yang diselesaikan.

### D. Forward Propagation

*Forward propagation* adalah proses meneruskan input melalui bobot dan operasi jaringan untuk memperoleh output. Bab ini menjadi dasar untuk memahami proses prediksi sebelum membahas mekanisme pembelajaran lebih lanjut.

## 3. Contoh Implementasi

Contoh berikut merupakan **ilustrasi konsep**, bukan klaim salinan kode asli buku. Kita menggunakan satu input dan satu bobot untuk menghitung prediksi.

```python
# Ilustrasi prediksi menggunakan satu input dan satu bobot

input_value = 8.5
weight = 0.1

prediction = input_value * weight

print("Input     :", input_value)
print("Bobot     :", weight)
print("Prediksi  :", prediction)
```

### Analisis Contoh

Nilai prediksi diperoleh dari perkalian input dengan bobot. Dengan input `8.5` dan bobot `0.1`, hasil prediksinya adalah `0.85`.

Contoh ini hanya menunjukkan cara kerja prediksi sederhana. Nilai tersebut belum menunjukkan akurasi model karena kita belum membandingkan prediksi dengan target sebenarnya.

```python
# Eksperimen: amati pengaruh perubahan bobot

weights = [0.1, 0.5, 1.0]

for weight in weights:
    prediction = input_value * weight
    print(f"Bobot = {weight:.1f} -> Prediksi = {prediction:.2f}")
```

## 4. Analisis Hasil

Pada input yang tetap, perubahan bobot menghasilkan perubahan prediksi. Hal ini menunjukkan bahwa bobot menentukan bagaimana input dipetakan menjadi output.

Eksperimen ini belum melakukan pelatihan dan belum mengukur kesalahan prediksi. Tujuannya hanya mengamati hubungan matematis antara input, bobot, dan output.

## 5. Kesimpulan

1. Prediksi merupakan tahap ketika jaringan saraf mengubah input menjadi output.
2. Bobot mengendalikan pengaruh input terhadap hasil prediksi.
3. Bentuk serta jumlah input dan output perlu disesuaikan dengan kebutuhan masalah.
4. Contoh satu bobot menjadi dasar untuk memahami jaringan yang lebih kompleks.
