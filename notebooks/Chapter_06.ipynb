## 1. Tujuan Bab

Bab ini memperkenalkan jaringan saraf yang memiliki lebih dari satu lapisan. Berbeda dengan model satu lapisan, jaringan dengan lapisan tersembunyi dapat mempelajari representasi perantara sebelum menghasilkan prediksi.

## 2. Konsep Teori

### A. Jaringan Saraf dengan Lapisan Tersembunyi

Jaringan saraf dapat memiliki input, satu atau beberapa lapisan tersembunyi, dan output. Setiap lapisan mengolah informasi menggunakan bobot.

### B. Forward Propagation

Forward propagation meneruskan input melalui jaringan hingga menghasilkan prediksi. Pada jaringan dua lapisan, output lapisan pertama menjadi input bagi lapisan berikutnya.

### C. Backpropagation

Backpropagation menghitung bagaimana error pada output berhubungan dengan bobot di dalam jaringan. Informasi gradien tersebut digunakan untuk memperbarui bobot agar prediksi berikutnya dapat membaik.

### D. Pembaruan Bobot

Gradient descent menggunakan gradien untuk mengubah bobot. Proses ini diulang selama pelatihan.

## 3. Implementasi Ilustratif

Contoh berikut memperlihatkan jaringan kecil dengan satu input, dua neuron tersembunyi, dan satu output. Kode ini merupakan ilustrasi mandiri, bukan reproduksi persis kode asli buku.

```python
import numpy as np

# Input dan target
x = np.array([[1.0]])
target = np.array([[0.8]])

# Bobot jaringan: input -> hidden -> output
W1 = np.array([[0.2, 0.4]])
W2 = np.array([[0.3], [0.5]])

learning_rate = 0.1

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(output):
    return output * (1 - output)

# Forward propagation
hidden = sigmoid(x @ W1)
prediction = sigmoid(hidden @ W2)

print("Aktivasi hidden:", hidden)
print("Prediksi awal:", prediction)
print("Target:", target)
```

```python
# Satu langkah backpropagation dan pembaruan bobot

error = prediction - target

# Gradien pada output
delta_output = error * sigmoid_derivative(prediction)
grad_W2 = hidden.T @ delta_output

# Gradien pada hidden layer
delta_hidden = (delta_output @ W2.T) * sigmoid_derivative(hidden)
grad_W1 = x.T @ delta_hidden

# Update bobot
W2 -= learning_rate * grad_W2
W1 -= learning_rate * grad_W1

# Prediksi setelah update
hidden_after = sigmoid(x @ W1)
prediction_after = sigmoid(hidden_after @ W2)

print("Prediksi sebelum update:", prediction.item())
print("Prediksi sesudah update:", prediction_after.item())
print("Squared error sesudah update:", ((prediction_after - target) ** 2).item())
```

## 4. Analisis Hasil

Forward propagation menghitung aktivasi pada lapisan tersembunyi dan menghasilkan prediksi. Backpropagation menghitung gradien dari output menuju lapisan sebelumnya, lalu memperbarui bobot.

Bandingkan prediksi dan squared error sebelum serta sesudah pembaruan. Satu langkah pembaruan tidak selalu menjamin error turun untuk semua konfigurasi, sehingga hasil aktual perlu diperiksa.
