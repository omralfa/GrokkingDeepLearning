## Tujuan bab
Bab ini menekankan cara membangun gambaran mental yang sederhana tentang jaringan saraf. Fokusnya bukan memperkenalkan algoritma pelatihan baru, melainkan menyederhanakan cara kita memikirkan korelasi, lapisan, bobot, dan aliran prediksi.

## Ringkasan teori dari buku
- **Correlation summarization:** jaringan saraf mempelajari pola/korelasi antara masukan dan target.
- **Lapisan tersembunyi:** node pada lapisan tengah dapat membentuk representasi antara dari kombinasi masukan. Contoh konseptual buku: untuk mengenali kucing, satu piksel biasanya tidak cukup; kombinasi piksel dapat mewakili ciri seperti mata atau telinga.
- **Aliran prediksi:** masukan dikalikan dengan bobot menuju lapisan berikutnya. Hasil pada satu lapisan menjadi masukan bagi lapisan berikutnya.
- **Visualisasi sederhana:** melihat jaringan sebagai rangkaian transformasi membantu kita memahami kode yang sebenarnya berisi operasi matriks.

## Gambaran jaringan tiga lapisan
`layer_0 (input) → weights_0_1 → layer_1 (hidden) → weights_1_2 → layer_2 (output)`

Notasi ini mengikuti pola variabel yang digunakan pada bab-bab sebelumnya. Contoh berikut adalah ilustrasi kecil untuk memperjelas aliran prediksi, bukan salinan penuh kode buku.

```python
import numpy as np

# Dua fitur masukan, dua neuron tersembunyi, satu keluaran.
layer_0 = np.array([[1.0, 0.5]])
weights_0_1 = np.array([[0.2, -0.1],
                        [0.4,  0.3]])
weights_1_2 = np.array([[0.7],
                        [-0.2]])

layer_1 = np.maximum(0, layer_0 @ weights_0_1)  # ReLU
layer_2 = layer_1 @ weights_1_2

print("layer_0 (input):", layer_0)
print("layer_1 (hidden):", layer_1)
print("layer_2 (output):", layer_2)
```

## Cara membaca hasil
1. `layer_0` menyimpan fitur masukan.
2. `layer_0 @ weights_0_1` menghitung kombinasi berbobot untuk lapisan tersembunyi.
3. ReLU mengubah nilai negatif menjadi nol.
4. `layer_1 @ weights_1_2` menghasilkan prediksi akhir.
